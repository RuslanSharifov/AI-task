## TASK 2
---
> 1. Sadə xətti reqressiya üçün a və b əmsallarını riyazi düsturlarla hesablayın.
> 2. Alınan reqressiya tənliyini ekrana çıxarın.
> 3. b əmsalının iqtisadi mənasını izah edin.
> 4. Hər işçi üçün proqnozlaşdırılan əməkhaqqını hesablayın.

```python
x_mean = np.mean(x)
y_mean = np.mean(y)

print("Orta təcrübə:", x_mean)
print("Orta əməkhaqqı:", y_mean)
```

```python
x_mean = np.mean(x)
y_mean = np.mean(y)

numerator = np.sum((x - x_mean) * (y - y_mean))
denominator = np.sum((x - x_mean) ** 2)

b = numerator / denominator

a = y_mean - b * x_mean

print("a (sərbəst hədd):", a)
print("b (meyl əmsalı):", b)
```

```python
y_pred = a + b * x




plt.scatter(x, y_pred, color='red', label='Əməkhaqqı')
plt.plot(x, y_pred, color='blue')
plt.xlabel("Iş təcrübəsi")
plt.ylabel("Əməkhaqqı")
plt.legend()
plt.show()
```

# Tapsiriq 3

> 1. LinearRegression modelini yaradın.
> 2. Modeli verilənlər üzərində öyrədin.
> 3. Modelin sərbəst həddini və meyl əmsalını ekrana çıxarın.
> 4. Alınan nəticələri riyazi üsulla hesablanan əmsallarla müqayisə edin.

```python
from sklearn.linear_model import LinearRegression




model = LinearRegression()

X = x.reshape(-1, 1)
model.fit(X, y)

a_model = model.intercept_
b_model = model.coef_[0]

print("Riyazi üsulla hesablanan əmsallar:")
print(f"a = {a:.2f}")
print(f"b = {b:.2f}")

print("\nScikit-learn ilə hesablanan əmsallar:")
print(f"a = {a_model:.2f}")
print(f"b = {b_model:.2f}")

# 7. Əmsalların eyni və ya çox yaxın olduğunu yoxlayaq
print("\nƏmsalların müqayisəsi:")
print("a eynidir:", np.isclose(a, a_model))
print("b eynidir:", np.isclose(b, b_model))

# 8. Reqressiya tənliyi
print(f"\nReqressiya tənliyi: y = {a_model:.2f} + {b_model:.2f} * x")
```

Hazırladığınız xətti reqressiya modelindən istifadə edərək aşağıdakı iş təcrübəsi müddətləri üçün əməkhaqqını proqnozlaşdırın.

### İnteraktiv proqramın hazırlanması

Proqramı elə dəyişdirin ki:

1. İstifadəçi klaviaturadan iş təcrübəsi müddətini daxil edə bilsin.
2. Proqram daxil edilmiş təcrübəyə uyğun proqnozlaşdırılan əməkhaqqını göstərsin.
3. İstifadəçi `0` daxil etdikdə proqram dayansın.
4. İstifadəçi ədəd əvəzinə mətn daxil etdikdə proqram xəta ilə dayanmasın.
5. Mənfi iş təcrübəsi müddətləri qəbul edilməsin.

```python
experiences = np.array([[2.5], [4.5], [7.5], [12]])

predictions = model.predict(experiences)

for experience, salary in zip(experiences.flatten(), predictions):
    print(f"{experience} il təcrübə → {salary:.2f} AZN")


# 2. İstifadəçidən təcrübə müddətinin alınması
while True:
    try:
        experience = float(input("\nİş təcrübəsini daxil edin (çıxış üçün 0): "))

        if experience == 0:
            print("Proqram dayandırıldı.")
            break

        if experience < 0:
            print("Xəta: Təcrübə müddəti mənfi ola bilməz.")
            continue

        salary = model.predict([[experience]])[0]

        print(f"Proqnozlaşdırılan əməkhaqqı: {salary:.2f} AZN")

    except ValueError:
        print("Xəta: Zəhmət olmasa, düzgün ədəd daxil edin.")
```
