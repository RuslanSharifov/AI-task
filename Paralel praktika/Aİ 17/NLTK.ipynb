# 1.  Tokenization

- ### 1.1 Word Tokenization ( Zözlərə görə tokenizasiya )


```python
text = """
Learning English is an important skill in today's world. It helps people communicate with others, find better jobs, and discover new opportunities. Every day, I try to learn new words, read interesting texts, and improve my writing skills. Sometimes learning a new language is difficult, but practice makes everything easier. I believe that patience and consistency are the keys to success. My goal is to become more confident and speak English fluently.
"""
print(text.split())
```

```python
from nltk.tokenize import word_tokenize
tokens = word_tokenize(text)
print(tokens)
```

- ### 1.2 Senrene Tokenization ( Cümlələrə görə tokenizasiya )

```python
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(text)

print("Sentences: ")
for i, sentences in enumerate(sentences, 1):
    print(f'     {i}. {sentences.strip()}')
```

# 1.3 Digər nov tokenizasiyalar

```python
from nltk.tokenize import TreebankWordTokenizer, WordPunctTokenizer, RegexpTokenizer
```

```python
text = "I cen't believe it! John's 25-years-old car costs $5,000.50."

treebank_tokenizer = TreebankWordTokenizer()
print("Treebank tokenizer: ")
print(treebank_tokenizer.tokenize(text))
```

```python
word_punct_tokenizer = WordPunctTokenizer()

print("Word Punctuation tokenizer: ")
print(word_punct_tokenizer.tokenize(text))
```

```python
regexp_tokenizer = RegexpTokenizer(r'\w+')
print("Regexp Tokenizer: ")
print(regexp_tokenizer.tokenize(text))
```

# 3. Lemmatization - lemmatizasiya

```python
from nltk.stem import WordNetLemmatizer

lammatizer = WordNetLemmatizer()

words = ['running', 'runs', 'ran', 'batter', 'geese', 'mice', 'wolves', 'cats']
print(('Nitg hisselerini gostermeden lammatizasiya (by default - isim'))

print('-' * 30)
for word in words:
    lamma = lammatizer.lemmatize(word)
    print(f'{word:15} =>  {lamma}')
```
