```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([45, 54, 64, 70, 75])
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
a = model.intercept_
b = model.coef_[0]
```

```python
print(f'Tenlik: y = {a:.2f} + {b:.2f}x')
```

```python
y_pred = model.predict(x)
y_pred
```

```python
plt.scatter(x, y, color='red', label='Datas')
plt.plot(x, y, color='blue', label='Regression')
plt.xlabel('Hours')
plt.ylabel('Marks')
plt.title('Simple Linear Regression')
plt.legend()
plt.show()
```

```python
while True:
    try:
        x_new = float(input('Enter a number: '))
        if x_new == 0:
            print('Exit')
            break
        y_new = a + b * x_new
        print(f'Predicted mark value: {y_new:.2f}')
        plt.scatter(x_new, y_new, color='red', label='out prediction', s=100)
        plt.scatter(x, y, color='red', label='datas')
        plt.plot(x, y, color='blue', label='Regression')
        plt.xlabel('Hours')
        plt.ylabel('Marks')
        plt.title('Simple Linear Regression')
        plt.legend()
        plt.show()
    except ValueError:
        print('Please enter a number')
```
