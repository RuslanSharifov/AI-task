## 7 gün üçün temperatur məlumatları əsasında xətti qrafik
--
### Verilənlər:
>temperature = [18, 20, 22, 21, 24, 26, 23]

> days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

### Tələblər

- plt.plot() istifadə edərək qrafik qurun.
- X oxunda günlər, Y oxunda temperatur göstərilsin.
- Qrafikə başlıq əlavə edin.
- marker və fərqli xətt stili istifadə edin.
- grid əlavə edin.
- Hər nöqtənin üzərində temperatur dəyərini göstərin.
- figsize ilə qrafikin ölçüsünü dəyişin.


```python
import secrets

import matplotlib.pyplot as plt
```

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]
```

```python
plt.figure(figsize=[10,5])

plt.plot(days, temperature, marker='o', linestyle='--')

plt.ylim(16, 30)
plt.grid(True)

plt.xlabel("Days")
plt.ylabel("Temperature")
plt.title("weekly temperature")

for i, temp in enumerate(temperature):
    plt.text(i, temp + 0.3, f'{str(temp)}C', ha="center", va="bottom")

plt.show()
```

# Proqramlaşdırma dilləri üzrə tələbə saylarını horizontal bar chart ilə göstərin.

## erilənlər:
> languages = ["Python", "JavaScript", "Java", "C#", "C++"]

> students = [35, 28, 22, 18, 15]

## Tələblər:

- plt.plot() istifadə edərək qrafik qurun.
- Hər bar üçün fərqli rəng istifadə edin.
- X və Y oxlarına ad verin və başlıq əlavə edin.
- Hər barın sonunda tələbə sayını plt.text() ilə göstərin.
- Ən çox tələbəsi olan dili vizual olaraq vurğulayın.
- Qrafikin ölçüsünü figsize ilə dəyişin.

## Əlavə
- Eyni məlumatı ayrıca plt.bar() istifadə edərək vertical bar chart kimi də göstərin.



```python
import numpy as np

languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]
```

```python
plt.figure(figsize=[10,5])

colors = ["red", "blue", "green", "orange", "purple"]
colors[0] = "gold"

plt.barh(languages, students, color=colors)

plt.title("Students by Programming Languages")
plt.ylabel("Languages")
plt.xlabel("Number of students")

plt.grid(axis="x")

for i, count in enumerate(students):
    plt.text(count + 0.3, i, str(count), va="center")

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(languages, students, color=colors)

plt.xlabel("Languages")
plt.ylabel("Number of students")
plt.title("Students by Programming Languages")

for i, count in enumerate(students):
    plt.text(i, count + 0.3, str(count), ha="center")

plt.show()
```

# Dərs saatı, imtahan nəticəsi və yaş məlumatlarını scatter qrafikində analiz edin.

## Verilənlər

> study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]

> scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]

> ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]



## Tələblər
- fig = plt.figure() və fig.add_axes(...) istifadə edin.
- study_hours və scores arasında scatter qrafiki yaradın.
- Nöqtələrin rəngi ages dəyişənindən asılı olsun və cmap istifadə edin
- marker və s parametrlərindən istifadə edin.
- colorbar əlavə edin və Age yazın.
- X oxu Study Hours, Y oxu Exam Score olsun.
- grid və uyğun başlıq əlavə edin.
- Ən yüksək nəticəli nöqtəni ayrıca qeyd edin.


## Əlavə

- Eyni Figure daxilində ikinci kiçik Axes yaradın və scores məlumatını bar qrafiki kimi göstərin.






```python
study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]

scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]

ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]
```

```python
fig = plt.figure(figsize=[10,5])

axes1 = fig.add_axes((0, 0, 1, 1))


scatter = axes1.scatter(
    study_hours,
    scores,
    cmap="viridis",
    c=ages,
    marker="o",
    s=150
)


axes1.set_title("Study Hours vs Exam Scores")
axes1.set_xlabel("Study Hours")
axes1.set_ylabel("Exam Score")
axes1.grid(True)

axes1.set_ylim(min(scores) - 5, max(scores) + 5)
axes1.set_xlim(0, 10)


max_index = scores.index(max(scores))
axes1.scatter(
    study_hours[max_index],
    scores[max_index],
    color="red",
    marker="*",
    s=300
)


axes2 =  fig.add_axes((0.05, 0.63, 0.3, 0.3))
axes2.bar(study_hours, scores)

axes2.set_xlim(0, 10)

axes2.set_title("Scores")

fig.colorbar(scatter, ax=axes1, label="Age")

```
