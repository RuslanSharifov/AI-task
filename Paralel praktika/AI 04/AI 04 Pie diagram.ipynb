```python
import matplotlib.pyplot as plt
```

```python
labels = ['Baku', 'Xirdalan', 'Absheron']
percents =[50, 35, 45]
```

```python
plt.pie(percents, labels=labels)
plt.title('Pie Diagram')
plt.show()
```

```python
plt.pie(percents, labels=labels, autopct="%1.2f%%")
plt.title('Pie Diagram')
plt.show()
```

```python
explode = (0.0, 0.05, 0)
plt.pie(
    percents,
    labels=labels,
    autopct="%1.2f%%",
    explode=explode,
    shadow=True,
    colors=["red", "green", "blue"],
    startangle=-50
)

plt.title('Pie Diagram')
plt.show()
```
