```python

import matplotlib.pyplot as plt
```

```python

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
temperature = [18, 20, 22, 21, 24, 26, 23]

plt.figure(figsize=(10, 5))
plt.plot(days, temperature, marker="o", linestyle='--')

plt.title("Weekly temperature")
plt.xlabel("Days")
plt.ylabel("temperature (^C)")
plt.grid(True)

plt.show()
```

```python
languages = ["Python", "JavaScript", "Java", "C#", "C++"]
students = [35, 28, 22, 18, 15]

plt.figure(figsize=(12, 6))
plt.title("Students by Programming Language")
plt.xlabel("Nubber of students")
plt.ylabel("Languages")
plt.barh(languages, students)

plt.show()

```

```python

study_hours = [1, 2, 2, 3, 4, 4, 5, 6, 7, 8]
scores = [45, 50, 55, 60, 65, 70, 74, 80, 88, 95]
ages = [18, 19, 18, 20, 21, 19, 22, 20, 23, 21]



fig = plt.figure(figsize=(10, 6))
ax = fig.add_axes([0.1, 0.1, 0.8, 0.8])

ax.scatter(
    study_hours,
    scores,
    c=ages,
    cmap="viridis",
    marker="o",
    s=150
)

scatter = ax.scatter(study_hours, scores, c=ages, cmap="viridis")

cbar = plt.colorbar(scatter, ax=ax)
cbar.set_label("Age")

ax.set_title("Study Hours vs Exam Score")
ax.set_xlabel("Study Hours")
ax.set_ylabel("Exam Score")
ax.grid(True)

ax.annotate(
    "Highest score",
    xy=(8, 95),
    xytext=(6, 92),
    arrowprops={"arrowstyle": "->"}
)

small_ax = fig.add_axes([0.14, 0.64, 0.25, 0.22])

small_ax.bar(range(1, 11), scores, color="steelblue")
small_ax.set_title("Scores", fontsize=9)
small_ax.tick_params(labelsize=7)
small_ax.grid(axis="y", alpha=0.2)


plt.show()

```
